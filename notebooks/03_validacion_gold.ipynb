```sql
SELECT 'ventas_sucursal' AS tabla, COUNT(*) AS registros
FROM electrocasa.gold.gold_ventas_sucursal

UNION ALL

SELECT 'ventas_mensuales', COUNT(*)
FROM electrocasa.gold.gold_ventas_mensuales

UNION ALL

SELECT 'ventas_categoria', COUNT(*)
FROM electrocasa.gold.gold_ventas_categoria

UNION ALL

SELECT 'satisfaccion_producto', COUNT(*)
FROM electrocasa.gold.gold_satisfaccion_producto

UNION ALL

SELECT 'devoluciones_producto', COUNT(*)
FROM electrocasa.gold.gold_devoluciones_producto

UNION ALL

SELECT 'motivos_devolucion', COUNT(*)
FROM electrocasa.gold.gold_motivos_devolucion

UNION ALL

SELECT 'tracking_courier', COUNT(*)
FROM electrocasa.gold.gold_tracking_courier

UNION ALL

SELECT 'dotacion_sucursal', COUNT(*)
FROM electrocasa.gold.gold_dotacion_sucursal;
```
