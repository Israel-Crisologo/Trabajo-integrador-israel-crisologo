```sql
%sql
SELECT 'ventas' AS tabla, COUNT(*) AS registros
FROM electrocasa.silver.silver_ventas

UNION ALL

SELECT 'productos', COUNT(*)
FROM electrocasa.silver.silver_productos

UNION ALL

SELECT 'empleados', COUNT(*)
FROM electrocasa.silver.silver_empleados

UNION ALL

SELECT 'resenas', COUNT(*)
FROM electrocasa.silver.silver_resenas

UNION ALL

SELECT 'devoluciones', COUNT(*)
FROM electrocasa.silver.silver_devoluciones

UNION ALL

SELECT 'tracking', COUNT(*)
FROM electrocasa.silver.silver_tracking;
```
